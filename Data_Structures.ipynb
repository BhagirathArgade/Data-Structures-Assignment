# Data Types And Structures Question

1. What are data structures, and why are they important?
* Data structures are ways to organize and store data (like lists, sets, dictionaries). They are important because they make it easier to manage and use data efficiently.

2. Explain the difference between mutable and immutable data types with examples.

* Mutable: Can be changed after creation (e.g., list → [1,2,3] → you can add/remove elements).

* Immutable: Cannot be changed after creation (e.g., string → "hello" → you cannot change letters directly).

3. What are the main differences between lists and tuples in Python?

*List → Mutable, written as [ ]

*Tuple → Immutable, written as ( )

4. Describe how dictionaries store data.
* Dictionaries store data as key-value pairs. Example: {"name":"Alex", "age":20}. The key is like a label, and the value is the data.

5. Why might you use a set instead of a list in Python?
* Because sets automatically remove duplicates and are faster for checking if an item exists.

6. What is a string in Python, and how is it different from a list?
* A string is a sequence of characters like "hello".
* A list is a sequence of items (numbers, strings, etc.) like [1, "hi", 3].

7. How do tuples ensure data integrity in Python?
* Tuples are immutable. Once created, their values cannot be changed. This prevents accidental modifications.

8. What is a hash table, and how does it relate to dictionaries in Python?
* A hash table is a data structure that maps keys to values quickly. Python dictionaries are built using hash tables.

9. Can lists contain different data types in Python?
* Yes, lists can store multiple data types. Example: [1, "hi", 3.5, True].

10. Explain why strings are immutable in Python.
* Strings are immutable so that they are safe and consistent. Once created, the same string cannot be changed, only replaced.

11. What advantages do dictionaries offer over lists for certain tasks?

*Faster searching by key

*Better for storing labeled data (like a phonebook: {"John":12345})

12. Describe a scenario where using a tuple would be preferable over a list.
* When you want to store data that should not change, like coordinates (10, 20) or days of the week.

13. How do sets handle duplicate values in Python?
* Sets automatically remove duplicate values. Example: {1,2,2,3} → {1,2,3}.

14. How does the “in” keyword work differently for lists and dictionaries?

In a list: checks if a value exists → 2 in [1,2,3] → True

In a dictionary: checks if a key exists → "name" in {"name":"Alex"} → True

15. Can you modify the elements of a tuple? Explain why or why not.
* No, tuples are immutable. Once created, you cannot change or remove their elements.

16. What is a nested dictionary, and give an example of its use case?
* A dictionary inside another dictionary. Example:

student = {"name":"Alex", "marks":{"math":90, "science":85}}


17. Describe the time complexity of accessing elements in a dictionary.
* Accessing a value by key is very fast (O(1)) on average because of hash tables.

18. In what situations are lists preferred over dictionaries?
* When you only need to store items in order and don’t need key-value pairs (like a shopping list).

19. Why are dictionaries considered unordered, and how does that affect data retrieval?
* Dictionaries don’t guarantee that items stay in the order you added them (before Python 3.7). So, you can’t rely on the order when retrieving.

20. Explain the differences between a list and a dictionary in terms of data retrieval.

* List → Access by position (index) → list[0]

* Dictionary → Access by key → dict["name"]
